```python
import pm4py
import os
import subprocess
from concurrent.futures import ProcessPoolExecutor, as_completed
import psutil
from pathlib import Path
import time
import gc
datasets_folder_path = "datasets"
root_dir = Path(datasets_folder_path)

scripts = list(root_dir.rglob("*.xes"))
```

```python
def wait_for_memory(threshold=0.5, check_interval=5): 
    """ Attende finché la RAM libera non supera la soglia indicata. threshold = 0.5 significa 50% di RAM libera. """ 
    time.sleep(check_interval * 2)  # Attesa iniziale per evitare controlli troppo frequenti
    while True: 
        mem = psutil.virtual_memory()
        free_ratio = mem.available / mem.total 
        if free_ratio >= threshold: 
            print(f"RAM libera sufficiente ({free_ratio:.2%}). Procedo.") 
            return 
        else: 
            print(f"RAM libera insufficiente ({free_ratio:.2%}). Aspetto {check_interval}s...")
            time.sleep(check_interval)
```

```python
def run(file_path):
    print(file_path)
    wait_for_memory(threshold=0.6)
    print(f"Inizio elaborazione: {file_path}")
    
    # Nota: wait_for_memory qui potrebbe essere problematico in parallelo.
    # Se molti processi partono insieme, potrebbero saturare la RAM prima di accorgersene.
    # wait_for_memory(threshold=0.5) 


    # Log su file per non intasare la RAM99
    log_name = f"log_{os.path.basename(file_path)}.txt"
    print("File di log", log_name)

    cmd_comand = ["python", "-u", "Complexity.py", "-f", str(file_path), "-m", "all"] 
    
    try:
        with open(log_name, "w") as log_file:
            result = subprocess.Popen(
                cmd_comand,
                stdout=log_file, 
                stderr=log_file,
                # check=True,
                text=True
            )
        print( f"OK: {file_path}")
    except subprocess.CalledProcessError as e:
        print( f"FAIL: {file_path} (Check {log_name})")
    except Exception as e:
        print( f"CRITICAL: {file_path} - {str(e)}")
```

```python
def cc(file_path):
    print(file_path)
    wait_for_memory(threshold=0.6, check_interval=10)
    print(f"Inizio elaborazione: {file_path}")
    
    # Nota: wait_for_memory qui potrebbe essere problematico in parallelo.
    # Se molti processi partono insieme, potrebbero saturare la RAM prima di accorgersene.
    # wait_for_memory(threshold=0.5) 


    # Log su file per non intasare la RAM
    log_name = f"cc_log_{os.path.basename(file_path)}.txt"
    print("File di log", log_name)

    cmd_comand = ["python", "-u", "CC.py", "-f", str(file_path)] 
    
    try:
        with open(log_name, "w") as log_file:
            result = subprocess.Popen(
                cmd_comand,
                stdout=log_file, 
                stderr=log_file,
                # check=True,
                text=True
            )
        print( f"OK: {file_path}")
    except subprocess.CalledProcessError as e:
        print( f"FAIL: {file_path} (Check {log_name})")
    except Exception as e:
        print( f"CRITICAL: {file_path} - {str(e)}")
```

```python
for elem in os.listdir():
    if elem.startswith("log") and elem.endswith(".txt"):
            with open(elem, "r") as f:
                dataset_name = str(elem).split("log_")[1].replace(".txt", ".xes")
                print(dataset_name, "*"*10)
                path = None
                for s_temp in scripts:
                    if dataset_name in str(s_temp):
                        path = str(s_temp)
                if path == None:
                    print(dataset_name, "Not found")
                    continue
                else:
                    print(dataset_name, "Found Found")
                                
                last_line = f.read().split("\n")[-2]
                
                if "Total" in last_line:
                        print(elem , "Ended")
                else:
                    print("RUN")
                    run(path)
                print("CC")
                cc(path)
```
